```python
import math
binomial_model = {
    "type": "binomial",
    "n": 10, #number of trials
    "p": 0.2 #probability of success
 }

poisson_model = {
    "type": "poisson",
    "rate": 2
 }

def pmf(model, k):   #probability mass function
    if model["type"] == "binomial":
        n = model["n"]
        p = model["p"]
        return math.comb(n, k) * p**k * (1-p)**(n-k)
    elif model["type"] == "poisson":
        rate = model["rate"]
        return (rate**k / math.factorial(k)) * math.exp(-rate)

def mean(model):
  if(model["type"] == "binomial"):
    return model["n"] * model["p"]
  elif(model["type"] == "poisson"):
    return model["rate"]

def variance(model):
  if(model["type"] == "binomial"):
    return model["n"] * model["p"] * (1 - model["p"])
  elif(model["type"] == "poisson"):
    return model["rate"]

models = [binomial_model, poisson_model]
for model in models:
  print(
    model["type"],
    "Probability:", pmf(model,5),
    "Mean:", mean(model),
    "Variance:", variance(model)
  )
```

```python
class DiscreetDistribution:
  def pmf(self,k):
    raise NotImplementedError

  def mean(self):
    raise NotImplementedError

  def variance(self):
   raise NotImplementedError
```

```python
class Binomial(DiscreetDistribution):
  def __init__(self,n,p):
     if not isinstance(n, int) or n<=0:
      raise ValueError("n must be a positive integer")
     if not (0<=p<=1):
      raise ValueError("p must be between 0 and 1")
     self.n = n
     self.p = p

  def pmf(self,k):
   if not isinstance(k, int) or k<0:
    raise ValueError("k must be a non-negative integer")
   return math.comb(self.n, k) * self.p**k * (1-self.p)**(self.n-k)

  def mean(self):
   return self.n * self.p

  def variance(self):
    return self.n * self.p * (1 - self.p)

```

```python
class Poisson(DiscreetDistribution):
  def __init__(self, rate):
    if rate<=0:
      raise ValueError("rate must be a positive number")
    self.rate = rate

  def pmf(self,k):
    if not isinstance(k, int) or k<0:
      raise ValueError("k must be a non-negative integer")

  def mean(self):
    return self.rate

  def variance(self):
    return self.rate
```

```python
def print_report(distribution, k):
  name = type(distribution).__name__

  print(f"\n{name}")
  print(f"Probability: {distribution.pmf(k)}")
  print(f"Mean: {distribution.mean()}")
  print(f"Variance: {distribution.variance()}")
```

```python
models =[
    Binomial(n=10, p = 0.2),
    Poisson(rate = 2)
]

for model in models:
  print_report(model, k=2)
```
